```python
from scipy.optimize import fsolve
import numpy as np

def gibbsfun(x, P, T):
    T_K = T + 273.15
    Keq = 10 ** (-12.275 + 4938 / T_K)

    muinCO = 100
    muinH2 = 600
    muinCH3OH = 0

    xi = x[0]

    muCO = muinCO - muinCO * xi
    muH2 = muinH2 - 2 * muinCO * xi
    muCH3OH = muinCH3OH + muinCO * xi
    mutot = muCO + muH2 + muCH3OH

    num = muCH3OH / mutot
    den1 = muCO / mutot
    den2 = (muH2 * muH2) / (mutot * mutot)

    return [num / (den1 * den2) - (P ** 2) * Keq]

T = 300
P = 200
xi0 = [0.5]
xi_solution = fsolve(gibbsfun, xi0, args=(P, T))
print(f"Conversion (xi): {xi_solution[0]}")
```
